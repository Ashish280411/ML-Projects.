```python
x=df.drop(columns=['price'],axis=1)
y=df['price']
```

```python
x
```

```python
y
```

```python
x_one_encoded=pd.get_dummies(x,columns=['model','transmission','fuelType'],drop_first=True)
```

```python
x_one_encoded
```

```python
x_one_encoded=x_one_encoded.astype(int)
```

```python
x_one_encoded
```

```python
from sklearn.preprocessing import LabelEncoder
columns=['model','transmission','fuelType']
x_label_encoded = x.copy()
label_encoder = {}
for col in columns:
  le=LabelEncoder()
  x_label_encoded[col]=le.fit_transform(x_label_encoded[col].astype(str))
  label_encoder[col]=le
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x_one_encoded, y, test_size=0.33, random_state=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred=model.predict(X_test)
```

```python
y_pred
```

```python
y_test
```

```python
r2=r2_score(y_test,y_pred)
r2
```

```python
n=X_test.shape[0]
p=X_test.shape[1]
adjusted_r2=1-(1-r2)*(n-1)/(n-p-1)
print("adjusted R^2 score:",adjusted_r2)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x_label_encoded, y, test_size=0.33, random_state=42)
```

```python
model12= LinearRegression()
model12.fit(X_train,y_train)
```

```python
y_pred=model12.predict(X_test)
```
